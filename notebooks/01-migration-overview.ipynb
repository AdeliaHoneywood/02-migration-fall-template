# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?
2. How might climate change affect animal migrations?

1. Migration happens across taxa in the animal kingdom. We are all familiar with bird migration and the monarch butterfly, and salmon migration is well-known. But other taxa migrate as well. When I took a School for Field Studies course in Africa we discussed how elephants historically migrated across the continent but modern human settlement has restricted thier movement. Hooved ungulates throughout the world are also known to migrate (Aikens et al., 2020). In Sequeira et al., 2025, we learned that marine megafauna including whales, sharks, penguins, polar bears, seals, sirenians, and turtles all migrate. What I never really thought about was that there are different types of migration (Fudickar et al., 2021). Many bird species are champs of obligate migration which occurs at the same time each year because  during the breeding season are not adequate in the winter. Irruptive migration is when an entire population moves because resources have been exhausted (Fudickar et al., 2021). The elephants from my Africa course were irruptive Because entire populations would move to greener pastures once they had eaten all of the vegetation in a place. Another type is facultative migration where individuals move in response to changing conditions (Fudickar et al., 2021). 

2. The chart of anthropogenic impacts in our reading by Sequeira et al., 2025 shows that climate change can impact migration of marine megafauna through changes in sea surface temperature that decrease prey population numbers or cause prey to change horizontal or vertical movement patterns. For migrating birds, the timing of spring green up is changing with climate change while birds' migration is more closely synced with historical timing of green-up rather than current conditions (Robertson et al., 2024). Terrestrial animals that migrate North, such as birds, insects, bats, and hooved ungulates are facing depleted conditions in their northern ranges because climate change is reducing food availability, providing opportunities for novel pathogens and parasites, and causing increased predation on offspring (Kubelka et al., 2022). Furthermore, bird species that winter in the southern hemisphere are declining faster then those that either do not migrate or only migrate within the northern hemisphere (Bairlein 2016, Rushing et al., 2020).

3. As a bird watcher, migration can structure my life and I pay a lot of attention to it. In the spring and fall we are watching for all kinds of songbirds including warblers and sparrows. In late summer the shorebirds come through. In the winter it is raptors and ducks, and finches if we are lucky. Different types of birds come through at different times and we plan our bird watching trips and destinations accordingly. Irruptive migration, or irruptions as we call them, can be  fun for birders such as winters where large numbers of snowy owls show up in the five boroughs, especially in Brooklyn, Queens, and Staten Island. Or when a variety of finch species come down in winter from the northern boreal regions. Both of these types of irruptions occur because of reductions in food in their northern habitat.
